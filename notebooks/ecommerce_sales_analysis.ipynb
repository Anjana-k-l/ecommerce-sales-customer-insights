# E-Commerce Sales & Customer Insights Dashboard
Portfolio Data Analyst project using Python, Pandas and SQL.

```python
import pandas as pd
from pathlib import Path
DATA = Path('../data/ecommerce_sales.csv')
df = pd.read_csv(DATA, parse_dates=['order_date'])
df.head()
```

```python
df.info()
df.isna().sum()
df.describe(include='all')
```

```python
kpis = {
    'orders': df.order_id.nunique(),
    'customers': df.customer_id.nunique(),
    'revenue': df.sales.sum(),
    'profit': df.profit.sum(),
    'profit_margin_pct': df.profit.sum()/df.sales.sum()*100,
    'avg_order_value': df.groupby('order_id').sales.sum().mean()
}
kpis
```

```python
df['month'] = df.order_date.dt.to_period('M').astype(str)
monthly = df.groupby('month', as_index=False).agg(revenue=('sales','sum'), profit=('profit','sum'))
category = df.groupby('category', as_index=False).agg(revenue=('sales','sum'), profit=('profit','sum'), orders=('order_id','nunique'))
region = df.groupby('region', as_index=False).agg(revenue=('sales','sum'), profit=('profit','sum'), orders=('order_id','nunique'))
monthly.head(), category.sort_values('revenue', ascending=False), region.sort_values('revenue', ascending=False)
```

## Business Takeaways
- Compare monthly revenue and profit trends to identify growth or slowdown periods.
- Use category and regional performance to prioritize inventory and marketing.
- Use repeat-customer analysis to identify retention opportunities.
- The Streamlit dashboard in `dashboard.py` provides an interactive view of the same KPIs.
